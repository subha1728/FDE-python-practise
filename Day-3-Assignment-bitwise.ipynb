#Left shift Right shift

```python
#Left shift
x=15
x<<=1
print(x)
#before binary 1111
#after binary 11110
#left shift (<<): move bits left and add bit position on right
```

```python
x=19   #10011
x<<=2  #10011000
print(x)
```

```python
#try
x=21
x<<=3
print(x)
```

```python
# right shift
x=15
x>>=1
print(x)

# >>1
#before x=15 1111
#after 0111
#right shift >> move bits right and discard the right most bit
```

```python
# right shift
x=15
x>>=2
print(x)

# >>1
#before x=15 1111
#after 0111
```

```python
#Left shift
15        1111 
15 << 1   11110
15 << 2   111100
15 << 3   1111000
```

```python
#right shift
15        1111 
15 >> 1   0111
15 >> 2   0011
15 >> 3   0001
```
